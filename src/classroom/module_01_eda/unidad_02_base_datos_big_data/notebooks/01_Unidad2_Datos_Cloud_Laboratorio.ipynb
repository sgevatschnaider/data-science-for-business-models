## Objetivos
- Reconocer fuentes de datos: IoT, ERP, CRM y API.
- Diferenciar datos estructurados, semiestructurados y no estructurados.
- Relacionar tipo/frecuencia/uso con decisiones de almacenamiento.
- Comparar IaaS, PaaS, SaaS y DaaS.

```python
import pandas as pd
import json
```

## 1. Datos estructurados

```python
ventas = pd.DataFrame({'cliente_id':[101,102,103],'canal':['web','tienda','web'],'importe':[120.5,89.0,210.2]})
ventas
```

La tabla posee columnas definidas y un esquema relativamente estable. Es natural consultarla con SQL o trabajarla como DataFrame.

## 2. Datos semiestructurados

```python
evento = {
 'device_id':'iot-77',
 'timestamp':'2026-08-19T10:15:00',
 'metrics': {'temperature': 22.7, 'humidity': 51},
 'tags':['planta_1','linea_A']
}
print(json.dumps(evento, indent=2))
```

JSON mantiene organización mediante claves y jerarquías, pero permite estructuras más flexibles que una tabla relacional.

## 3. Un registro no estructurado

```python
texto = 'El cliente indica que el producto llegó tarde pero valora la atención recibida.'
print(texto)
```

El texto libre no tiene columnas predefinidas. Para analizarlo habría que extraer rasgos, entidades, sentimiento u otras representaciones.

## 4. Comparación de fuentes

```python
fuentes = pd.DataFrame([
 ['IoT','sensores/eventos','alta','telemetría'],
 ['ERP','operaciones internas','media','stock/finanzas'],
 ['CRM','clientes/interacciones','media','ventas/marketing'],
 ['API','intercambio entre sistemas','variable','servicios externos']
],columns=['fuente','origen','frecuencia','ejemplo'])
fuentes
```

## 5. Modelos de servicio cloud

```python
cloud = pd.DataFrame([
 ['IaaS','infraestructura','alto','servidores/VMs'],
 ['PaaS','plataforma','medio','desarrollo/despliegue'],
 ['SaaS','software','bajo','aplicación lista'],
 ['DaaS','datos','bajo','datos externos preparados']
],columns=['modelo','qué_consumo','control_técnico','foco'])
cloud
```
