## Objetivos
- Diferenciar aprendizaje supervisado, no supervisado y por refuerzo.
- Relacionar regresión, clasificación, clustering y reducción dimensional con preguntas concretas.
- Ejecutar ejemplos mínimos y comparar qué cambia entre paradigmas.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression, make_classification, make_blobs
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import mean_absolute_error, accuracy_score
```

## 1. Supervisado · Regresión

```python
X,y=make_regression(n_samples=180,n_features=1,noise=18,random_state=42)
m=LinearRegression().fit(X,y)
pred=m.predict(X)
print('MAE:',round(mean_absolute_error(y,pred),2))
plt.scatter(X[:,0],y,alpha=.55); plt.plot(np.sort(X[:,0]),m.predict(np.sort(X,axis=0))); plt.title('Regresión'); plt.show()
```

## 2. Supervisado · Clasificación

```python
Xc,yc=make_classification(n_samples=250,n_features=4,n_informative=3,n_redundant=0,random_state=7)
clf=LogisticRegression(max_iter=1000).fit(Xc,yc)
pc=clf.predict(Xc)
print('Accuracy:',round(accuracy_score(yc,pc),3))
```

## 3. No supervisado · Clustering

```python
Xu,_=make_blobs(n_samples=260,centers=3,cluster_std=1.2,random_state=12)
km=KMeans(n_clusters=3,n_init=10,random_state=42).fit(Xu)
plt.scatter(Xu[:,0],Xu[:,1],c=km.labels_); plt.title('K-Means: estructura sin etiquetas'); plt.show()
```

## 4. No supervisado · Reducción dimensional

```python
Xp=np.c_[Xc, Xc[:,0]*.5 + np.random.default_rng(1).normal(0,.2,len(Xc))]
pca=PCA(n_components=2).fit(Xp)
Z=pca.transform(Xp)
print('Varianza explicada acumulada:',round(pca.explained_variance_ratio_.sum(),3))
plt.scatter(Z[:,0],Z[:,1],c=yc,alpha=.6); plt.title('Representación PCA en 2D'); plt.show()
```

## 5. Aprendizaje por refuerzo · idea mínima

```python
rng=np.random.default_rng(4)
true_rewards=[0.2,0.5,0.8]
est=np.zeros(3); n=np.zeros(3)
for t in range(300):
    a = rng.integers(3) if rng.random()<0.15 else np.argmax(est)
    r = rng.normal(true_rewards[a],0.2)
    n[a]+=1; est[a]+= (r-est[a])/n[a]
print('Recompensas estimadas:',np.round(est,3))
print('Acción preferida:',int(np.argmax(est)))
```

## Matriz de decisión
- **Valor continuo conocido:** regresión.
- **Clase conocida:** clasificación.
- **Sin etiqueta y buscamos grupos:** clustering.
- **Sin etiqueta y buscamos representación compacta:** reducción dimensional.
- **Decisiones secuenciales con recompensa:** aprendizaje por refuerzo.

## Cierre
El paradigma de ML se elige a partir de la pregunta y de la evidencia disponible. El algoritmo es una decisión posterior.
